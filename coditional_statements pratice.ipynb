```python
i = 1
while i<=5:
    print(' datascience', end = "") # when we mention end then new line will not create
    j = 1
    while j<=4:
        print(' technology', end= "")
        j = j + 1
        
    i = i + 1
    print()
```

```python
# what is nested if 
#checks one condition and then check inside it check another"
#first condition must be true 
# then second condition is checked 

age = 20
has_id = True 

if age >= 18:
    if has_id:
        print("allowed entry")
```

```python
# difference between elif vs nested if 
# elif sperates the conditions only one runs 
#only one block will excute 
  

x = 10 

if x > 10:
    print("greater")
elif x == 10:
    print("equal")    
```

```python
# nested if dependent conditions 
# second depends on first

x = 10 
 
if x >= 10:
    if x == 10:
        print("exactly 10") 
```

```python
# if elif else 
#only one can exist at a time 
# if you want multiple outcomes dont use elif 

signal = "red"

if signal == "green":
    print("go")
elif signal == "yellow":   
    print("slow down")  
elif signal == "red":
    print("stop")
else:
    print ("invalid signal")       
```
